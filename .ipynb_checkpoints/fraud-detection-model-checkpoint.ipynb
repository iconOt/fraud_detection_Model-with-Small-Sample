```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from imblearn.over_sampling import SMOTE
from sklearn.pipeline import Pipeline
import joblib
import random

# 1) create syncthetic dataset
def generate_demo_data(n_samples=20000, fraud_rate=0.02, random_state=42):
    np.random.seed(random_state)
    random.seed(random_state)

    # Amount: heavy-tailed distribution (many small, same large)
    amounts = np.round(np.random.exponential(scale=50, size=n_samples), 2)     # e.g., mean ~50

    # Time: seconds since midnight (0 - 86399)
    times = np.random.randint(0, 86400, size=n_samples)

    # Transaction types and locations
    transaction_types = np.random.choice(['ATM', 'POS', 'Online'], size=n_samples, p=[0.4, 0.45, 0.15])
    locations = np.random.choice(['Abuja', 'Lagos', 'Kano', 'Others'], size=n_samples, p=[0.25, 0.45, 0.15, 0.15])

    # Create a friend probability model (so fraud have patterns)
    probs = np.full(n_samples, 0.005) # baseline low fraud prob
    # increase probability for large amounts
    probs += (amounts > 200).astype(float) * 0.02
    # Online transactions slightly riskier
    probs += (transaction_types == 'Online').astype(float) * 0.03
    # odd hours (e.g., midnight - 4am) slightly riskier
    probs += ((times < 4*3600) | (times > 20*3600)).astype(float) * 0.01
    # location risk tweak
    probs += (locations == 'others').astype(float) * 0.005

    # Randomly sample labels from the probabilities
    labels = (np.random.rand(n_samples) < probs).astype(int)

    df = pd.DataFrame({
        'amount': amounts,
        'time': times,
        'transaction_type': transaction_types,
        'location': locations
        'class': labels
    })
    return df

df = generate_demo_data(n_samples=20000)
print("Dataset shape:", df.shape)
print(df['class'].value_counts(normalize=True))

# 2) Encode categorical features to march your Flask mappings
transaction_map = {'ATM': 0, 'POS': 1, 'Online': 3}
location_map = {'Abuja': 0, 'Lagos': 1, 'Kano': 2, 'Others': 3}

df['transaction_type'] = df['transaction_type'].map(transaction_map)
df['location_code'] = df['location'].map(location_map)

# 30 Prepare X,y
X= df[['amount', 'time', 'transaction_type_code', 'location_code']].values
y = df['class'].values

# 4) Train/test split
X_train, X_test, y_train, y_test = train_train_split(X,  y, stratify=y, test_size=0.2, random_state=42)

# 5) Handle imbalance with SMOTE (optional) or use class_weight
sm = SMOTE(random_state=42)
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
print("After SMOTE class distribution:", np.bincount(y_train_res))

# 6) Building pipeline: scaling + classifier
pipeline = Pipline ([
    ('scaler', StandardScaler()),
    ('clf', RandomForestClassifier(n_estimators=150, random_state=42, class_weight='balanced'))
])

pipeline.fit(x_train_res, y_train_res)

# 7) Evaluate
y_pred = pipeline.predict(X_test)
y_proba = pipeline.predict_proba(X_test)[:, 1]

print(classification_report(y_pred))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, y_proba))

# 8) Save the pipepline (includes caler + model) to fraud_model.pkl
joblib.dump(pipline, "fraud_model.pkl")
print("Saved model to fraud_model.pkl")

# Quick test: load and predict a sample
model = joblib.load("fraud_model.pkl")
sample = np.array([[5000.0,45000, transaction_map['POS'], lacation_map['Lagos']]]) # amount, time, type,location
Print(y_train.value_counts())
```
